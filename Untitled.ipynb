```python
import psycopg2
from sqlalchemy import create_engine
```

```python
DATABASE = 'equities'
USER = 'postgres'
PASSWORD = 'root'
HOST= 'localhost'
PORT = '5432'

connect_alchemy = "postgresql+psycopg2://%s:%s@%s/%s" % (
    USER,
    PASSWORD,
    HOST,
    DATABASE
)
engine = create_engine(connect_alchemy)
# conn = psycopg2.connect(database=DATABASE,user=USER,password=PASSWORD,host=HOST,port=PORT)
# cur = conn.cursor()
```

```python
df = pd.read_csv("HINDALCO_1D.csv", parse_dates=[0])
```

```python
df.head()
```

```python
df.to_sql(name='prices', con=engine, if_exists='replace', index=False)
```

```python
conn = psycopg2.connect(database=DATABASE,user=USER,password=PASSWORD,host=HOST,port=PORT)
cur = conn.cursor()
```

```python
cur.execute("SELECT column_name, data_type FROM information_schema.columns WHERE table_name = 'prices'")
```
